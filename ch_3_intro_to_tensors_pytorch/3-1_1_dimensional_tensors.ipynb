```python
v = torch.tensor([1, 2, 3, 4, 5, 6])
print(v)
```

```python
print(v.dtype)
```

```python
print(v[0])
print(v[1])
print(v[2])
```

```python
print(v[:])
```

```python
print(v[1:4]) # stop index is exclusive
```

```python
print(v[1:])
```

```python
print(f.dtype)
```

```python
print(f.size())
```

```python
a = np.array([1, 2, 3, 4, 5])
print(a)
```

```python
tensor_cnv = torch.from_numpy(a)
print(tensor_cnv)
print(tensor_cnv.type())
```

```python
numpy_cnv = tensor_cnv.numpy()
print(numpy_cnv)
```
