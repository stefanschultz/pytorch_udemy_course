```python
import torch
```

```python
# as example following function
# 9x⁴+2x³+3x²+6x+1

# Deriative
# (d f(x)) / dx

x = torch.tensor(2.0, requires_grad=True)
y = 9*x**4 + 2*x**3 + 3*x**2 + 6*x + 1
y.backward()
x.grad
```

```python
# y = x²+z³

# (d / (d*x)) * (z³ + x²) = 2x

# (d / (d*z)) * (z³ + x²) = 3z²

# f'(x, z) = f'(1, 2)

x = torch.tensor(1.0, requires_grad=True)
z = torch.tensor(2.0, requires_grad=True)
y = x**2 + z**3
y.backward()
x.grad
```
